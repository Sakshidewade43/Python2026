### 2. Write a Python program to read first n lines of a file.


```python
f = open('sample.txt','w')
f.write("Hello , I am sakshi, \nAnd I am a Data Science Engineer")
f.close()



n= int(input("Enter number of lines to read: "))
with open('sample.txt','r') as f:
  for i in range(n):
    line = f.readline()
    if not line:   #stop if file ends before n lines
      break
    print(line,end="")
```

### 9. Write a Python program to count the frequency of words in a file.


```python
with open('sample.txt','r') as f:
  word_count = {}

  for line in f:
    words = line.strip().split() #split line into words

    for word in words:
      if word in word_count:
        word_count[word]+=1
      else:
        word_count[word]=1
    print("word frequency:")
    for word,count in word_count.items():
      print(word ,":", count)
```
